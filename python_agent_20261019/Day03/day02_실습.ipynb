### pandas 설치하기

```python
# 임포트하기
import pandas as pd
```

### page6.chapter1.DataFrame 만들기와 기본 속성

```python
data = {
    "name": ["철수", "영희", "민수"],
    "score": [88, 95, 72]
}
df = pd.DataFrame(data)

print(df.shape)      # (3, 2)  -> 행 3개, 열 2개
print(df.columns)    # Index(['name', 'score'])
print(df.index)      # RangeIndex(start=0, stop=3)
```

### page7.chapter1.[실습] 나만의 DataFrame 만들기

```python
# [1번 문제]
# 아래 정보를 딕셔너리로 만들고 DataFrame으로 변환하세요.
# 상품명(item): 마우스, 키보드, 모니터
# 가격(price): 15000, 45000, 220000
data = {
    "item":['마우스','키보드','모니터'],
    "price":[15000, 45000, 220000],
}

df = pd.DataFrame(data)
# [2번 문제]
# df.shape 와 df.columns 를 출력해
# 행/열 개수와 컬럼명을 확인하세요.
print(df.shape)
print(df.columns)
# [도전 문제]
# df.shape[0] 을 이용해 '총 상품 수: 3개' 형태로 출력해보세요
print(f"총 상품 수: {df.shape[0]}개") # (3,2)
```
