### pandas 설치하기

```python
# 임포트하기
import pandas as pd
```

### page6.chapter1.DataFrame 만들기와 기본 속성

```python
data = {
    "name": ["철수", "영희", "민수"],
    "score": [88, 95, 72]
}
df = pd.DataFrame(data)

print(df.shape)      # (3, 2)  -> 행 3개, 열 2개
print(df.columns)    # Index(['name', 'score'])
print(df.index)      # RangeIndex(start=0, stop=3)
```

### page7.chapter1.[실습] 나만의 DataFrame 만들기

```python
# [1번 문제]
# 아래 정보를 딕셔너리로 만들고 DataFrame으로 변환하세요.
# 상품명(item): 마우스, 키보드, 모니터
# 가격(price): 15000, 45000, 220000
data = {
    "item":['마우스','키보드','모니터'],
    "price":[15000, 45000, 220000],
}

df = pd.DataFrame(data)
# [2번 문제]
# df.shape 와 df.columns 를 출력해
# 행/열 개수와 컬럼명을 확인하세요.
print(df.shape)
print(df.columns)
# [도전 문제]
# df.shape[0] 을 이용해 '총 상품 수: 3개' 형태로 출력해보세요
print(f"총 상품 수: {df.shape[0]}개") # (3,2)
```

page9.chapter2.[실습] CSV 읽고 쓰기

```python
data = {"item": ["연필", "지우개"], "price": [500, 300]}
df = pd.DataFrame(data)

# [1번 문제]
# df를 'items.csv' 파일로 저장하세요. (index=False 옵션 사용)
df.to_csv("items.csv", index=False)

# [2번 문제]
# 저장한 파일을 pd.read_csv() 로 다시 읽어 출력하세요.
df_csv = pd.read_csv("items.csv")
print(df_csv)

# [도전 문제]
# df를 'items.json' 파일로도 저장하고 다시 읽어보세요.
# force_ascii=False(한글처리), indent(줄바꿈)
df.to_json("items.json",force_ascii=False,indent=4)
df_json=pd.read_json("items.json")
```

### page12.chapter3.[실습] 데이터 탐색과 필터링

```python
data = {
    "name": ["철수", "영희", "민수", "지현"],
    "score": [88, 95, 60, 72],
}
df = pd.DataFrame(data)

# [1번 문제] df.describe() 로 score 컬럼의 통계를 확인하세요.
print(df.describe())

# [2번 문제] score가 80 이상인 행만 필터링해 출력하세요.
print(df[df['score']>=80])
print(type(df['score']))

# [도전 문제] name 컬럼만 선택해 리스트로 바꿔보세요. (.to_list())
print(type(df['name'].to_list()))
print(df['name'].to_list())
```

### page14.chapter4.[실습] 결측치와 중복 처리

```python
data = {
    "name": ["철수", "영희", "영희", "민수"],
    "score": [88, 95, 95, None],
}
df = pd.DataFrame(data)

# [1번 문제] df.isna().sum() 으로 결측치 개수를 확인하세요.
print(df.isna().sum())

# [2번 문제] fillna(0) 으로 결측치를 채운 결과를 출력하세요.
print(df.fillna(0))
df=df.fillna(0) # 원본에 다시 대입해야 반영됨
print(df)

# [도전 문제] drop_duplicates() 로 중복된 행을 제거해보세요.
df=df.drop_duplicates()
print(df)
```
