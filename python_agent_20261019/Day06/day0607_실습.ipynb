### page9.chapter2.[실습]PromptTemplate + Chain으로 응답 생성기

```python
from langchain_core.prompts.prompt import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm2 = ChatOpenAI(
    model="gpt-4o-mini", #모델 이름
    temperature=0.9,       #창의적인 답변 정도
    max_tokens=300,      #답변 토큰량
    api_key=api_key,     #api key
)

# [1번 문제]
# "{food}의 유래를 한 문장으로 알려줘" 템플릿을 PromptTemplate으로 만드세요.
template = PromptTemplate.from_template(
    "{food}의 유래를 한 문장으로 알려줘"
)
# [2번 문제]
# 템플릿 | llm | StrOutputParser() 로 체인을 만들고, food="김치"로 invoke()해서 결과를 출력하세요.
parser = StrOutputParser()
chain = template | llm2 | parser
result = chain.invoke({"food":"김치"})
print(result)

# [도전 문제]
# food 값을 "피자", "초밥" 등으로 바꿔가며 여러 번 호출해보고, 매번 결과가 조금씩 달라지는지 확인하세요.
result2 = chain.invoke({"food":"피자"})
print(result2)
result3 = chain.invoke({"food":"초밥"})
print(result3)
```

page12.chapter3.[실습] Tool 두 개 만들어보기

```python
from langchain_core.tools import tool

#[1번 문제] 사칙연산 문자열(예: "3*4+2")을 계산해 반환하는 calculate(expression: str) 함수를 @tool로 작성하세요.
# eval("3*4") : 보안상 위험하기 때문에 사용을 권장하지 않는다
# LLM이 결과를 받아서 사용해야되니깐 항상 문자열로 결과를 반환하자--> str()
@tool
def calculate(expression:str) -> str:
    """사칙연산 계산식 문자열을 계산합니다. 예) '3*4+2' """
    return str(eval(expression))

#[2번 문제] 단어를 받아 글자 수를 반환하는 count_letters(word: str) 함수를 @tool로 작성하세요.
@tool
def count_letters(word: str) -> str:
    """단어를 받아 글자 수를 반환합니다. """
    return str(len(word))

#[도전 문제] 두 Tool의 .name과 .description을 각각 출력해서, 
# docstring이 description으로 그대로 들어가는지 확인하세요.
print(f"calculate:{calculate.name},descrip:{calculate.description}")
print(calculate.invoke({'expression':'3*4+2'}))
print(count_letters.invoke({'word':'안녕하세요'}))



```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

from langchain_core.messages import AIMessage
from langchain_core.messages import HumanMessage
prompt = ChatPromptTemplate.from_messages([
    ("system", "너는 친절한 상담원이야."),
    MessagesPlaceholder(variable_name="history"),   # 이전 대화 기록이 들어갈 자리
    ("human", "{question}"),
])
history = []   # 대화가 쌓일수록 이 리스트에 추가해나감

chain = prompt | llm | StrOutputParser()

question = "당신은 누구인가요?"

# 질문을 보낼때 과거 대화기록도 같이 보내자 -> LLM이 이전 대화를 기억한다.
result = chain.invoke({"history":history,"question":question})

print(result)

# 질문, 답변 내용을 기록해줘야 함
history.append(HumanMessage(content=question)) #사용자 질문
history.append(AIMessage(content=result)) #AI 답변

question2 = "내가 방금 뭐라고 질문 했지?"

# 질문을 보낼때 과거 대화기록도 같이 보내자 -> LLM이 이전 대화를 기억한다.
result = chain.invoke({"history":history,"question":question2})
print(result)
```

### page15.chapter4.[실습] Memory를 가진 대화형 체인 만들기

```python
from langchain_core.messages import HumanMessage
from langchain_core.messages import AIMessage

# [1번 문제] MessagesPlaceholder("history")를 포함한 ChatPromptTemplate으로 체인을 만드세요.
promptTemplate = ChatPromptTemplate.from_messages([
    ('system','너는 친절한 상담원이야'),
    MessagesPlaceholder(variable_name="history"),
    ('human','{question}')
])

# [2번 문제] history=[] 상태에서 "내 이름은 하윤이야"를 invoke()하고, history에 사용자/AI 메시지를 각각 추가하세요.
history=[] # 이전 대화내용을 저장할 저장객체

chain1=promptTemplate | llm | parser

question="내 이름은 하윤이야"
# chain에 history와 질문을 함께 전달해서 호출한다.
result = chain1.invoke({"history":history,'question':question})
print("AI답변:",result)

# history에 사용자 질문과 AI답변을 저장한다.
history.append(HumanMessage(content=question))
history.append(AIMessage(content=result))

# [도전 문제] 이어서 같은 history로 "내 이름이 뭐라고 했지?"를 invoke()해서, 답변에 이름이 포함되는지 확인하세요.
question = "내 이름이 뭐라고 했지?"
result2 = chain1.invoke({"history":history,'question':question})
print("AI답변:",result2)
```

### page17.chapter4.[실습] 종합 : Tool + Memory를 가진 페르소나 챗봇

```python
# [1번 문제] calculate(expression) Tool을 다시 활용해, system 메시지에 "너는 친절한 수학 선생님이야" 페르소나를 가진 
# ChatPromptTemplate을 만드세요.
prompt = ChatPromptTemplate.from_messages([
    ('system',"너는 친절한 수학 선생님이야"),
    MessagesPlaceholder(variable_name='history'),
    ('human','{question}')
])
# [2번 문제] MessagesPlaceholder로 history를 연결한 체인을 만들고, "3 * 4 + 2는 얼마야?"를 invoke()하세요.
calcu1="3*4+2"
chain = prompt | llm | StrOutputParser()

# 대화기록 리스트
history_17 = []

# tool calculate를 먼저 이용해서 결과를 얻어온다.
calcu_result1 = calculate.invoke({'expression':calcu1})
print('Tool 실행결과:',calcu_result1)

#chain에 넣어줄 질문을 계산결과와 함께 문장으로 만든다.
q1=f"{calcu1}의 계산결과는 {calcu_result1}야. 곱셈을 덧셈보다 먼저해야 하는 이유와 함께 친절하게 설명해줘."

#AI에게 질문 전달하고, 결과 받기
result1 = chain.invoke({'history':history_17,'question':q1})

print("AI답변:",result1)

#대화 기록 저장
history_17.append(HumanMessage(content=q1))
history_17.append(AIMessage(content=result1))

# [도전 문제] 이어서 "방금 계산한 값에 10을 더하면?"을 같은 history로 invoke()해서, 이전 계산 결과를 기억한 채 답하는지 확인하세요.
# 직전 계산 결과를 가져와 10을 더한다.
calcu2 = f"{calcu_result1} + 10"
calcu_result2 = calculate.invoke({'expression':calcu2})

q2 = f"방금 계산한 값에 10을 더하면 {calcu_result2}가 됨. 맞게 계산했는지 확인해줘"

#AI에게 질문 전달하고, 결과 받기
result2 = chain.invoke({'history':history_17,'question':q2})

print("AI답변:",result2)

#대화 기록 저장
history_17.append(HumanMessage(content=q2))
history_17.append(AIMessage(content=result2))

# Tool의 역할: LLM에 연산을 맡기지 않고, calculate로 정확히 계산해서 처리
# Memory의 역할: 계산 부담없이 친절한 수학 선생님 컨셉과 이전 대화 문맥을 이어 가는 설명에 집중하게 함
# 추후 Agent가 내부적으로 Tool과 LLM을 어떻게 조합해서 쓰는지 이해할 수 있음
```
