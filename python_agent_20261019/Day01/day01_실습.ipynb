### page10.chapter2.[실습] 계산기 + 조건식 통합

```python
price = 12000
quantity = 5
total = price * quantity # 총가격

# [1번 문제]
# 결제 금액(total)이 50000 이상인지 확인하는 비교식(>=)을 작성하세요.
# 그 결과값(True 또는 False)을 is_free_shipping 이라는 변수에 저장하세요.
is_free_shipping= total >= 50000
print('무료배송여부:',is_free_shipping)

# [2번 문제]
# 무료배송(is_free_shipping) 대상이거나, 구매 수량(quantity)이 10개 이상인지
# 확인하세요. 둘 중 하나라도 만족하면 True가 나오도록 논리 연산자(or)를
# 사용하고, 그 결과를 is_vip 변수에 저장하세요.
is_vip = is_free_shipping or (quantity >=10)
print('VIP 여부:',is_vip)
# 여기에 코드를 작성해보세요
```

### page15.chapter3.[실습] 문자열 실전 - 이메일 정리기

```python
raw = "  hong@example.com  "

# [1번 문제]
# raw 문자열의 앞뒤에 붙은 공백을 strip() 메서드로 제거하고,
# 그 결과를 email 이라는 변수에 저장하세요.
email=raw.strip()
print(len(email))

# [2번 문제]
# email을 split("@") 으로 나누면 ['hong','example.com'] 처럼
# 아이디와 도메인이 담긴 리스트가 됩니다.
# 이 리스트를 user_id, domain 두 변수에 나눠 담으세요.
splitVal = email.split("@")
user_id=splitVal[0]
domain=splitVal[1]

# [3번 문제]
# f-string으로 아래 형태로 출력하세요.
# '아이디: hong / 도메인: example.com'
val = f"아이디:{user_id} / 도메인:{domain}"
print(val)

# [도전 문제]
# replace() 메서드로 domain의 'example'을 'school'로 바꿔 출력해보세요.
domain = domain.replace('example','school')
print(f"아이디:{user_id} / 도메인:{domain}")

```

### page18.chapter3.[실습] 리스트 실전 - 점수 분석

```python
scores = [88, 92, 75, 92, 60]

# [1번 문제]
# scores 리스트에서 92점이 몇 번 나왔는지 count() 메서드로 확인하세요.
count = scores.count(92)
print(f"count:{count}")

# [2번 문제]
# scores.sort() 로 점수를 오름차순 정렬한 뒤 결과를 출력하세요.
scores.sort()
print(f"sort:{scores}")

# [3번 문제]
# enumerate()를 이용해 순서를 붙여 아래 형태로 출력하세요.
# '1번째 학생: 88점'
for i , score in enumerate(scores):
    print(f"{i}번째 학생:{score}점")
# [도전 문제]
# min(), max()로 최고점과 최저점을 각각 구하고, 그 차이를 계산해 출력하세요.
score_max = max(scores)
score_min = min(scores)
print(f"최고점, 최저점:{score_max},{score_min}")
```

### page22.chapter3.[실습] 종합 - 상품 목록 분석

```python
products = [
    {"name": "마우스", "price": 15000},
    {"name": "키보드", "price": 45000},
    {"name": "모니터", "price": 220000},
]

# [1번 문제]
# products 안의 상품명(name)만 꺼내 리스트로 만들어 출력하세요.
# 예: ['마우스', '키보드', '모니터']
names=[]
for product in products:
    names.append(product["name"])
print(names)

# [2번 문제]
# 각 상품의 가격(price)이 20000원 이상인지 비교연산자(>=)로 확인해
# 각각 True/False로 출력하세요.
for product in products:
    print(product["name"],"20000원 이상:",product["price"]>=20000)

# [3번 문제]
# 세 상품의 가격을 모두 더해 전체 합계를 계산해 출력하세요.
sum =0
for product in products:
    sum+=product["price"]
print("총 가격:",sum)

```
