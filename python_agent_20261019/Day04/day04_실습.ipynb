### page6.chapter1.[실습] Numpy 배열 다루기

```python
import numpy as np

scores = np.array([88, 92, 75, 60, 95])

# [1번 문제]
# scores 배열의 평균(mean), 최댓값(max), 최솟값(min)을
# 각각 출력하세요.
print(scores.mean())
print(scores.max())
print(scores.min())
# [2번 문제]
# scores 배열에서 80점 이상인 값만 필터링해 출력하세요.
scores_80 = scores[scores >= 80]
print(scores_80)
# [도전 문제]
# 모든 값에 5점을 더한 새 배열을 만들어 출력하세요.
scores_5=scores+5
print(scores_5)
```

page10.chapter2.[실습] 기본 그래프 그리기

```python
months = ["1월", "2월", "3월", "4월", "5월"]
visitors = [120, 150, 90, 200, 170]

# [1번 문제]
# months를 x축, visitors를 y축으로 하는 라인 그래프를 그리세요.

# [2번 문제]
# title(), xlabel(), ylabel()로 제목과 축 이름을 추가하세요.

# [도전 문제]
# color="orange", marker="s" 옵션을 추가해 꾸며보세요.

#한글 폰트 및 마이너스 부호 설정
plt.rc('font',family='Malgun Gothic')
plt.rc('axes',unicode_minus=False)

plt.plot(months,visitors,
         color="orange",marker="s") # s, ^, v, <, >, o 등등
plt.title('방문자현황')
plt.xlabel('월')
plt.ylabel('방문자수')
```

### page13.chatper3.[실습] Seaborn으로 시각화하기

```python
import seaborn as sns

scores = [55, 68, 72, 80, 85, 90, 95, 60, 77, 83]

# [1번 문제]
# sns.histplot() 으로 scores의 분포를 히스토그램으로 그리세요.
sns.histplot(scores)
plt.savefig('check_13a.png')
plt.close()
plt.show()

# [2번 문제]
# sns.boxplot() 으로 scores의 박스플롯을 그리세요.
sns.boxenplot(y=scores)
plt.savefig('check_13b.png')
plt.close()
plt.show()
# [도전 문제]
# 임의의 hours(공부 시간) 리스트를 만들고 sns.scatterplot()으로
# hours와 scores의 관계를 산점도로 그려보세요.
hours=[1,2,3,4,5,6,7,4,3,5]
sns.scatterplot(x=hours,y=scores)
plt.savefig('check_13c.png')
plt.close()
plt.show()
```
